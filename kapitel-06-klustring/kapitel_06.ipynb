# Kapitel 6 - Klustring

## Faktafrågor

### 1. Vad är klustring för något? Ge några exempel på tillämpningsområden.

Klustring betyder att man delar upp data i olika grupper, där data som liknar varandra hamnar i samma grupp. Man behöver alltså inte bestämma grupperna själv från början, utan modellen försöker hitta grupperna åt oss.

Klustring kan till exempel användas för att dela upp kunder efter deras köpbeteende, hitta olika typer av bostäder eller dela upp musik efter olika egenskaper.

### 2. Förklara översiktligt hur *K-means* fungerar. Använd figur 6.3 (sidan 238) och figur 6.4 (sidan 239) i din förklaring. 

K-means delar upp data i olika grupper. Först bestämmer man hur många grupper man vill ha. Sedan försöker modellen hitta mitten på varje grupp. Därefter läggs varje punkt i den grupp som den ligger närmast. Modellen flyttar sedan mitten på grupperna och gör om detta flera gånger tills grupperna blir så bra som möjligt.

I figur 6.3 ser man exempel på hur olika kluster kan se ut. Figur 6.4 visar att K-means inte alltid fungerar bra, till exempel när klustren har olika form, spridning eller storlek.

## Resonemangfrågor

### 3. Hur kan man välja vilket antal kluster som ska användas för en K-means modell? Använd *inertia*, *silhouette score* och *silhouette diagram* i ditt svar. 

Man kan testa olika antal kluster och se vilket som fungerar bäst. 

**Inertia** visar hur nära punkterna ligger mitten av sitt kluster. Inertia blir lägre när man använder fler kluster, så man kan titta efter en punkt där förbättringen börjar bli mindre.

**Silhouette score** visar hur bra punkterna passar i sina kluster. Ett högre värde betyder oftast att klustren är bättre separerade.

Med ett **silhouette diagram** kan man se hur bra punkterna passar i sina kluster och lättare jämföra olika antal kluster. 

Man kan alltså använda både inertia och silhouette score för att få en bild av vilket antal kluster som passar bäst.

### 4. Om du kollar på figur 6.10 på sidan 247, hur många kluster hade du valt och varför? Är det en "exakt vetenskap" att välja antalet kluster?

Jag skulle välja **k=4 kluster**.

**Varför?**  
Grafen visar inertia för olika antal kluster. Från k=1 till k=4 sjunker kurvan mycket. Efter k=4 sjunker den mycket mindre. Det betyder att 4 kluster verkar vara en bra gränspunkt. Det finns en "armbåge" (elbow) där kurvan börjar plana ut.

**Är det exakt vetenskap?**  
Nej, det är det inte. Man kan välja olika antal kluster och få okej resultat. Det beror på vad man vill göra och vad som passar datan. Inertia-kurvan ger därför bara en vägledning, och man kan även använda till exempel silhouette score för att jämföra olika antal kluster.

### 5. Hur tolkar man figur 6.13 på sidan 251? 

Silhouette diagram visar hur bra varje punkt passar i sitt kluster. Längre staplar och högre värden betyder att punkterna passar bättre i sina kluster. Om en stapel går över till ett negativt värde betyder det att den punkten kanske hade passat bättre i ett annat kluster.

En viktig sak att kolla är om staplarna är ganska jämna och om klustren är ungefär lika stora. Man kan också kolla om de flesta staplar är relativt långa. Den röda streckade linjen visar genomsnittet för alla punkter.

Om man kollar på figurerna ser man att k=4 har jämnast klusterstorlekar och inga tydligt sämre kluster. k=3 fungerar också relativt bra, men ett av klustren är betydligt större än de andra två. k=5 och k=6 har mer varierande stapellängder, vilket tyder på att några kluster passar sämre.
