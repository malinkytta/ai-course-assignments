### 2. Vad menas med att en modell produktionssätts? 

Det betyder att modellen slutar vara ett experiment i till exempel en notebook och istället börjar användas på riktigt i en app, på en hemsida eller i något annat system där riktiga användare eller andra program får nytta av modellens prediktioner.

Exempelvis: istället för att bara köra `model.predict()` själv i Jupyter kopplas modellen in i en app som automatiskt föreslår priser till kunder varje dag.

### 3. Vad är *scikit-learn* för något? Biblioteket följer några centrala designprinciper. Vilka är dessa? Vad är *estimators*, *predictors* och *transformers*? 

Scikit-learn är ett Python bibliotek som innehåller flera verktyg för maskininlärning. Det finns bland annat färdiga modeller att träna, verktyg för att förbereda data och verktyg för att mäta hur bra en modell fungerar.

En viktig sak med scikit-learn är att olika verktyg följer samma enkla struktur. Tre viktiga begrepp är:

- **Estimator** - ett objekt som kan lära sig något från data genom metoden `.fit()`. Det är ett slags samlingsbegrepp för många objekt i scikit-learn.
- **Predictor** - en estimator som också kan göra förutsägelser med `.predict()`. En `LinearRegression` är till exempel både en estimator och en predictor.
- **Transformer** - en estimator som omvandlar data med `.transform()`, till exempel genom att skala siffror eller gör one-hot encoding. Ofta använder man `.fit_transform()`, som gör båda stegen på en gång.

Tack vare att verktygen följer samma mönster blir det enklare att byta ut en modell mot en annan utan att behöva skriva om en massa kod.

### 6. Många AI/ML projekt uppnår inte de ursprungligen satta målen eller att ens passera någon form av prototyp-stadie. Vad tror du detta beror på och hur ska vi förhålla oss till det? 

Det kan finnas flera olika orsaker:

- För dålig eller för lite data - man har inte tillräckligt bra eller tillräckligt mycket data för att modellen ska kunna lära sig något användbart
- Otydliga problem - man har inte förstått exakt vad man vill lösa innan man börjar bygga modellen
- Orealistiska förväntningar - man tror att AI kan lösa allt, men vissa problem passar helt enkelt inte för maskininlärning
- Svårt att gå från prototyp till produktion - att få en modell att fungera i en notebook är mycket enklare än att få den att fungera stabilt, snabbt och säkert i ett riktigt system med riktiga användare
- Organisatoriska problem - till exempel bristande samarbete mellan utvecklingsteam och resten av verksamheten, eller att ingen egentligen ansvarar för att underhålla modellen efteråt

Hur vi bör förhålla oss till det är att se AI/ML som ett experiment och inte som en garanterad lösning. Börja smått, testa idén tidigt och var beredd på att ändra strategi eller lägga ner projektet om det visar sig att det inte fungerar.

Precis som i fråga 1 arbetar man sällan i en rak linje i sådana projekt. Det är också viktigt att ha en tydlig bild av vad som faktiskt är möjligt att uppnå, så att förväntningarna på projektet är realistiska.

###  8. Förklara vad koden nedan gör. Varför är det viktigt att kunna spara en modell? 

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from joblib import dump, load

X, y = make_regression(n_samples=20000, n_features=3, noise=0.1)

model = LinearRegression().fit(X, y)
dump(model, "linear_model.joblib")
model_loaded = load("linear_model.joblib")

print(model_loaded.predict(X[:5]))
```


Steg för steg:

1. `make_regression(...)` skapar ett påhittat dataset med 20 000 rader och 3 features. Det används här bara för att ha data att träna modellen på.
2. `LinearRegression().fit(X, y)` skapar en linjär regressionsmodell och tränar den på datan.
3. `dump(model, "linear_model.joblib")` **sparar** den tränade modellen till en fil på datorn, med hjälp av biblioteket `joblib`.
4. `load("linear_model.joblib")` **laddar tillbaka** modellen från filen och sparar den i variabeln `model_loaded`
5. `model_loaded.predict(X[:5])` använder den inladdade modellen för att göra prediktioner på de fem första raderna i datasetet. Det visar att modellen fortfarande fungerar efter att den har sparats och laddats in igen.

**Varför är det viktigt att kunna spara en modell?**  
Att träna en modell kan ta allt från sekunder till dagar/veckor beroende på hur stor och komplex den är. Utan att kunna spara modellen skulle man behöva träna om den från början varje gång man vill använda den igen, vilket inte är särskilt praktiskt. 

Genom att spara modellen kan man till exempel:
- Ladda in den direkt i en app eller ett system utan att träna om den
- Dela modellen med andra
- Fortsätta jobba med modellen senare utan att behöba börja om från början
