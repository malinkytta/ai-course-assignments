# Kapitel 3 - Regression

## Faktafrågor

### 1. Vad kännetecknar regressionsproblem? Ge några exempel på tillämpningsområden.

Ett regressionsproblem är när vi vill gissa ett tal, inte en kategori. Talet kan bli nästan vad som helst, t.ex. 84 500 kr eller 12,7 grader.

Exempel:

- Gissa priset på en bil utifrån miltal och ålder.
- Gissa priset på ett hus utifrån storlek och antal rum.
- Gissa hur mycket potatis en åker ger, baserat på väder och gödsling.
- Gissa en persons lön utifrån utbildning och antal års erfarenhet.

### 2. Förklara utvärderingsmåtten RMSE, MSE och MAE.

Alla tre svarar på samma fråga: *"hur mycket missar modellen i snitt?"* Lägre värde = bättre modell.

- **MSE** – ta felet på varje gissning, multiplicera felet med sig själv (så stora fel blir mycket större), och ta medelvärdet av alla dessa.
- **RMSE** – samma som MSE, fast du tar roten ur svaret på slutet. Det gör att talet hamnar i samma enhet igen (t.ex. kronor, inte "kronor i kvadrat"), vilket gör det lättare att förstå.
- **MAE** – istället för att kvadrera felen gör du dem bara positiva (absolutbelopp) och tar medelvärdet.

Skillnad: RMSE straffar stora missar extra hårt. MAE bryr sig lika mycket om alla fel, stora som små.

### 3. Om vi ska rangordna olika modeller, spelar det någon roll om RMSE eller MSE används? Varför?

Nej. RMSE är bara MSE med roten ur, så ordningen mellan modellerna blir alltid densamma, vinner en modell med lägst MSE, vinner den även med lägst RMSE.

Enda skillnaden: MSE är lite snabbare att räkna ut (du slipper räkna ut roten), så om man testar väldigt många modeller kan MSE spara lite tid. Men RMSE används oftast ändå eftersom den är enklare att förstå/tolka.

### 4. Förklara mycket översiktligt vad gradient descent är.

Gradient descent är ett sätt för modellen att steg för steg hitta de bästa inställningarna, genom att minska felet lite i taget.

Tänk dig att du står högst upp på en kulle i tjock dimma och ska ta dig ner till botten (botten = minsta möjliga fel). Du känner vilket håll som lutar neråt, tar ett steg dit, känner igen, tar ett nytt steg och fortsätter tills marken känns helt platt. Då har du hittat botten.

Två saker som påverkar hur det går:

- **Learning rate/steg-storlek** – hur stora steg du tar. För stora steg och du kan missa botten helt. För små steg och det tar väldigt lång tid.
- **Antal iterationer** – hur många steg du max får ta innan du stannar.

### 5. Vad är the bias variance trade-off? Varför är mer komplexa modeller inte alltid bättre?

Felet en modell gör kan delas in i tre delar:

- **Bias** – felet som kommer av att modellen är för enkel/gör för enkla antaganden. T.ex. att dra en rak linje genom data som egentligen böjer sig. Ger underanpassning.

- **Variance** – felet som kommer av att modellen är för känslig och ändrar sig mycket beroende på exakt vilken träningsdata den fick. Ger överanpassning.

- **Irreducerbart fel** – slump/brus i datan som ingen modell kan göra något åt.

**Varför är mer komplexa modeller inte alltid bättre?** Ju mer avancerad en modell blir, desto bättre kan den fånga komplicerade mönster (lägre bias), men den blir samtidigt känsligare och mer *nyckfull* (högre variance). Vinsten äts alltså ofta upp av nackdelen. Bästa modellen är inte den mest avancerade, utan den med bäst balans mellan de här två.

### 6. Förklara översiktligt hur respektive modell fungerar.

a) **Linjär regression** – Ritar en rak linje genom datan som beskriver sambandet mellan x och y så bra som möjligt.

b) **Ridge regression** – Samma som linjär regression, men håller modellens vikter lite mindre, så den inte blir för känslig för träningsdatan.

c) **Lasso regression** – Liknar ridge, men kan sätta oviktiga variablers vikt till exakt noll, väljer alltså bort dem helt automatiskt.

d) **Elastic net** – En mix av ridge och lasso, där du själv bestämmer hur mycket av varje du vill ha.

e) **Support vector machines (SVM)**– Försöker hitta den bästa gränsen som skiljer två grupper åt.

f) **Beslutsträd** – Ställer en rad ja/nej-frågor om datan (t.ex. "är åldern över 30?") tills den hamnar i en grupp och kan läsa av ett svar. Lätt att förstå steg för steg.

g) **Ensemble learning** – Istället för en modell, använder man flera modeller och slår ihop deras gissningar (tex tar medelvärdet), i hopp om att det blir bättre tillsammans.

h) **Random forest** – Många beslutsträd som tillsammans röstar fram ett svar (medelvärdet av allas gissningar). Blir oftast stabilare än ett enda träd.

### 7. Vad menas med white box modeller och black box modeller?

- **White box-modeller** – man kan lätt se och förklara *varför* modellen gjorde en viss gissning. Beslutsträd är ett bra exempel, du kan följa frågorna steg för steg.

- **Black box-modeller** – modellen gissar ofta bra, men det är svårt att förklara i ord *varför* den kom fram till just det svaret. Random forest och neurala nätverk är exempel.

### 8. Vad är skillnaden mellan bagging och pasting?

Båda handlar om att skapa flera nya, lite olika dataset genom att slumpmässigt plocka observationer från det ursprungliga datasetet, och sen träna en modell på varje nytt dataset.

- **Bagging** – samma observation får förekomma flera gånger i samma nya dataset.
- **Pasting** – samma observation får bara vara med en gång.

### 9. Förklara hur man kan tolka figur 3.1 på sidan 113.

Bilden visar fyra saker:


- **Axlarna** – lodrätt är inkomst, vågrätt är ålder. Varje **svart prick** är en person: hur gammal den är och vad den tjänar.

- **Den blåa linjen** – modellens bästa gissning. Ju äldre man är, desto högre upp på linjen (mer inkomst), enligt modellen.

Långt rött streck = linjen gissade fel för den personen. Kort streck = linjen gissade nästan rätt.

### 11. Vad är determinationskoefficienten ($R^2$)?

$R^2$ är ett mått som visar hur stor del av variationen i $y$ som modellen lyckas förklara.

- $R^2 = 1$ betyder att modellen förklarar allt perfekt.
- $R^2 = 0$ betyder att modellen inte är bättre än att bara gissa medelvärdet varje gång.
- $R^2$ kan till och med bli negativt, om modellen är sämre än att bara gissa medelvärdet.

Till skillnad från RMSE, MSE och MAE (som är i en specifik enhet, t.ex. kronor) är $R^2$ bara ett tal mellan 0 och 1 (oftast), vilket gör det enklare att jämföra olika modeller rakt av, oavsett vad de mäter.
