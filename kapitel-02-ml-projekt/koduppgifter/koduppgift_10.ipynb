### 10. Datasetet "salary_dataset.csv" innehåller 29 observationer av personer. `YearsExperience` är hur många år de arbetat, och `Salary` är deras lön.

#### a) Läs in datasetet "salary_dataset.csv" med pandas `read_csv()`-funktion och dela upp i `X` och `y`. Avgör själv vilken variabel som ska vara den beroende variabeln `y`.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

from sklearn.metrics import mean_squared_error
```

```python
data_salary = pd.read_csv("../../data/salary_dataset.csv")
data_salary.info()
data_salary.head()
```

#### b) Dela upp datasetet i träning- och testset. (Inget validerings-set alltså!)

```python
X_salary = data_salary[["YearsExperience"]]
y_salary = data_salary[["Salary"]]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_salary, y_salary, test_size=0.2, random_state=42)
```

#### c) Träna två regressionsmodeller med k-delad korsvalidering med `cross_validate()`-funktionen från *scikit-learn*. Använd `neg_root_mean_squared_error` som `scoring`. Välj själv hur många iterationer den ska göra genom hyperparametern `cv`.


```python
from sklearn.model_selection import cross_validate

lin_reg_s = LinearRegression()
tree_reg_s = DecisionTreeRegressor(max_depth=3, random_state=42)

lin_cv = cross_validate(lin_reg_s, X_train_s, y_train_s, cv=4, scoring="neg_root_mean_squared_error")

tree_cv = cross_validate(tree_reg_s, X_train_s, y_train_s, cv=4, scoring="neg_root_mean_squared_error")

print("Linear Regression RMSE:", round(-lin_cv["test_score"].mean(), 2))
print("Decision Tree RMSE:", round(-tree_cv["test_score"].mean(), 2))
```

#### d) Utvärdera modellen som presterar bäst på testsetet.

```python
best_model = LinearRegression()
best_model.fit(X_train_s, y_train_s)

y_pred_s = best_model.predict(X_test_s)
test_rmse = np.sqrt(mean_squared_error(y_test_s, y_pred_s))

print("RMSE på testsetet:", round(test_rmse, 2))
```
