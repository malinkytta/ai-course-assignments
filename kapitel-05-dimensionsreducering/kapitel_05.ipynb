## Faktafrågor

### 1. Vad menas med *curse of dimensionality*

Det betyder att data kan bli svårare att jobba med ju fler dimensioner (variabler) man har.

När man lägger till fler dimensioner blir det mer plats för punkterna att sprida ut sig på. Datan blir då mer utspridd, och det blir svårare för modellen att hitta bra mönster i datan.

Man behöver också mer data för att modellen ska fungera bra. Det kan också göra att modellen tar längre tid att träna.

Det blir också svårare att visualisera datan. Man kan enkelt visa upp till tre dimensioner i en graf, men har man fler blir det svårt att visa direkt.

Man kan tänka på det som att leta efter något i ett rum. Ju större och rörigare rummet är, desto svårare blir det att hitta det man letar efter.


### 2. Vad är dimensionsreducering och varför görs det?

Dimensionsreducering betyder att man minskar antalet variabler eller dimensioner i datan.

Man försöker minska datan samtidigt som man behåller så mycket viktig information som möjligt.

Det gör man bland annat för att:
- modellen ska gå snabbare
- det ska bli enklare att jobba med datan
- minska risken för överanpassning
- göra det lättare att visa datan i exempelvis en graf

### 6. Efter att vi genomfört en *PCA*, vad händer med tolkningen av variablerna? 

Efter PCA blir det svårare att förstå vad de olika variablerna betyder.

Det beror på att PCA skapar nya variabler som kallas huvudkomponenter, till exempel C1 och C2.

De nya variablerna är en kombination av de gamla variablerna.

Så istället för att ha till exempel:

## Koduppgifter

### 8. Förklara vad nedanstående kod gör. 

```python
import numpy as np
from sklearn.decomposition import PCA

# Creating a dataset with 3 features/columns
X = np.random.rand(1000, 3)
print(X[0:5])
# Reducing the data to 2 dimensions
pca = PCA(n_components=2)
X2D = pca.fit_transform(X)
print(X2D[0:5])

# "Recreating" the data to 3 dimensions
X3D_inv = pca.inverse_transform(X2D)
# Not exactly equal since some information was lost in the transformation
print(np.allclose(X3D_inv, X))
```

Det börjar med att skapa ett dataset med 1000 rader och 3 variabler. Värdena skapas slumpmässigt mellan 0 och 1.

Sedan används PCA för att minska antalet dimensioner från 3 till 2. PCA försöker då behålla så mycket variation i datan som möjligt.

Efter det försöker koden göra om datan till 3 dimensioner igen med `inverse_transform`. Datan blir dock inte exakt likadan som från början eftersom en dimension togs bort och därför gick viss information förlorad.

Till sist används `np.allclose` för att kolla om den nya datan är tillräckligt lik originaldatan. Resultatet blir `False`, eftersom `X3D_inv` inte är tillräckligt lik originaldatan efter att en dimension tagits bort.
